```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

b3db = pd.read_csv("../data/processed/b3db_features.csv")
FEATURE_COLS = ["MolWt", "LogP", "TPSA", "NumHDonors", "NumHAcceptors", "NumRotatableBonds"]

print(b3db["label"].value_counts())
print(b3db["label"].value_counts(normalize=True))  # class imbalance as a percentage
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.flatten(), FEATURE_COLS):
    sns.boxplot(data=b3db, x="label", y=col, ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.savefig("../notebooks/descriptor_distributions.png")  # save for README
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(b3db[FEATURE_COLS].corr(), annot=True, cmap="coolwarm", center=0)
plt.title("Descriptor correlation matrix")
plt.savefig("../notebooks/correlation_heatmap.png")
plt.show()
```
